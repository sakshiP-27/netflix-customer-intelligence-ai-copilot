This is a fixed investigation, not a free-roaming agent. For one country and one month, Python runs the same five queries every time: churn versus the previous month and versus the whole base, watch time and sessions, support tickets and satisfaction, failed payments, and the plan mix.

The model sees only that evidence. It should say a change is consistent with another change, not that one caused the other. A region that is not in `customers.country` stops before any query or model call. `state_or_region` is empty in this extract, so the region is the country.

The two months below are inside the data. India in May 2026 and the United States in April 2026. Atlantis is not a country.
