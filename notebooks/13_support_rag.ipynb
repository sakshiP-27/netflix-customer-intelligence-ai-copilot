# Week 3 Step 5: What customers are saying

I do not ask the model what customers say. It has never read the tickets. I search the real ticket and feedback text, put the closest records in the prompt, and ask the model to summarize only those records.

The search index is saved under `data/vector_store`. The first run embeds every usable ticket and feedback entry, which downloads a small local model and takes a few minutes. Later runs reuse that index.

The record count at the end of each answer is added by the code, not by the model.


```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

from src.llm.support_rag import ask_support, build_support_index

index = build_support_index()
print(index)
```

```python
payment = ask_support("what are customers saying about payment failures?")
print("status:", payment["status"])
print(payment["answer"])
print()
for record in payment["records"][:5]:
    print(f"[{record['n']}] {record['source']} | {record['category']} | {record['text'][:180]}")
```

A question that the tickets cannot cover should say so. The count line still reports how many records were retrieved, because those records are a sample, not proof about every customer.
